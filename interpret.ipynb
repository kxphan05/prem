```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dataset = pd.read_csv('./mydata.csv')
dataset.head()
```

```python
X = dataset.drop(columns=['links', 'class', 'Goals Home', 'Away Goals'], axis=1)
y = pd.DataFrame((dataset['class'] == 'h') * 1)
categorical_X = [cname for cname in X.columns
                 if X[cname].dtype == 'object'
                 and len(X[cname].unique()) < 5]
numerical_X = [cname for cname in X.columns
               if X[cname].dtype in ['float64', 'int64']
               and len(X[cname].unique()) > 5]
print(categorical_X, numerical_X, y)
```

We see that there are no categorical variables (well we can use time or team as a categorical variable but it would be unnecessarily complex). I also removed away and home goals as it leads to test train contamination. We create a simple tree model.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

X = X[numerical_X]
X_train, X_val, y_train, y_val = train_test_split(X, y)
pipe = Pipeline(steps=[
    ('scaler', StandardScaler()),
    ('tree', RandomForestClassifier())
])
pipe.fit(X_train, y_train)
y_pred = pipe.predict(X_val)
print(accuracy_score(y_val, y_pred))
```
